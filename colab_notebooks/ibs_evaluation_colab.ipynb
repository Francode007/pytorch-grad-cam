## 2. Configure Evaluation Parameters

Adjust the parameters below to control the evaluation process.

```python
# --- Evaluation Parameters ---

# Model and Dataset
MODEL_NAME = "resnet18"  # e.g., "resnet18", "resnet50", "efficientnet_b4"
DATASET_TYPE = "ibs"
# IMPORTANT: Update this path to your dataset location in Google Drive or another accessible path
DATASET_PATH = '/content/drive/MyDrive/IBS_RESEARCH/dataset/IBS_dataset/IBS-preprocessed-dataset'

# Enhanced CAM Configuration
ENHANCED_CAM_METHOD = "GradCAMEnhanced"  # "GradCAMEnhanced", "GradCAMPlusPlusEnhanced", etc.
LAYER_MODE = "all"  # "last", "last_5", "all"

# Standard CAM Methods to Evaluate
STANDARD_METHODS = ["GradCAM", "GradCAM++", "ScoreCAM", "HiResCAM"]

# Evaluation Settings
MAX_IMAGES = -1  # Number of images to evaluate. Use -1 for the entire dataset.
STEP_SIZE = 50  # Step size for insertion/deletion curves.
DEVICE_PREFERENCE = "auto" # "auto", "cuda", "cpu"

# Output Directories
BASE_CSV_DIR = "./csv_exports"
BASE_ANALYSIS_DIR = "./analysis_results"
```

## 3. Run Evaluation

This cell will execute the full evaluation based on your configuration.

```python
from XAI_Enhancer_module.multi_dataset_xai_evaluation import MultiDatasetXAIEvaluationSuite

# Initialize the evaluation suite
eval_suite = MultiDatasetXAIEvaluationSuite(
    model_name=MODEL_NAME,
    dataset_type=DATASET_TYPE,
    device_preference=DEVICE_PREFERENCE,
    layer_mode=LAYER_MODE,
    enhanced_cam_method=ENHANCED_CAM_METHOD
)

# Update dataset path if needed (especially for Colab)
eval_suite.dataset_path = DATASET_PATH

# Print dataset info
eval_suite.print_dataset_info()

# Run the full evaluation
results = eval_suite.run_full_evaluation(
    methods=STANDARD_METHODS,
    max_images=MAX_IMAGES,
    include_enhanced=True,
    step_size=STEP_SIZE,
    base_csv_dir=BASE_CSV_DIR,
    base_analysis_dir=BASE_ANALYSIS_DIR
)

print("\\n✅ Evaluation complete.")
```

## 4. Display Results

The following cell will display the evaluation results and show where the output files are saved.

```python
import pandas as pd
from XAI_Enhancer_module.utils.directory_manager import print_directory_structure

print("="*50)
print("Evaluation Results Summary")
print("="*50)

if 'comparison_df' in results and results['comparison_df'] is not None:
    print("\\n📊 Comparison DataFrame:")
    display(results['comparison_df'])
else:
    print("\\nNo comparison DataFrame was generated.")

print("\\n📂 Final Directory Structure:")
print_directory_structure(base_analysis_dir=BASE_ANALYSIS_DIR, base_csv_dir=BASE_CSV_DIR)
```
